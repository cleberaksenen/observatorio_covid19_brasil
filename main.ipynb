```python
from pathlib import Path
```

```python
PROJETO = Path.cwd()
BRONZE = PROJETO / "Camada_Bronze"
SILVER = PROJETO / "Camada_Silver"

if not BRONZE.is_dir():
    raise FileNotFoundError(
        f"Camada_Bronze não encontrada em {PROJETO}. "
        "Abra o notebook na pasta que contém a Camada_Bronze."
    )

for subpasta in ["scripts", "dados", "qualidade", "documentacao"]:
    (SILVER / subpasta).mkdir(parents=True, exist_ok=True)

print(f"Estrutura criada em: {SILVER}")
```
